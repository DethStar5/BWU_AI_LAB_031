```python
from sklearn.datasets import load_digits
import matplotlib.pyplot as plt

digits = load_digits()

print("Dataset Shape:", digits.data.shape)
print("Image Shape:", digits.images.shape)
print("Target Labels:", digits.target_names)

plt.figure(figsize=(5, 1))
plt.imshow(digits.images[9], cmap='gray')
plt.title(f"Label: {digits.target[5]}")
plt.axis('off')
plt.show()
```

```python
from sklearn.datasets import load_digits
import matplotlib.pyplot as plt

digits = load_digits()

fig, axes = plt.subplots(2, 6, figsize=(10, 10))

for i, ax in enumerate(axes.flat):
    ax.imshow(digits.images[i], cmap='grey')
    ax.set_title(f"Digit: {digits.target[i]}")
    ax.axis('off')
    
plt.tight_layout()
plt.show()
```

```python
from sklearn.datasets import load_iris
import pandas as pd

iris = load_iris()

df = pd.DataFrame(iris.data, columns=iris.feature_names)

df["Target"] = iris.target

print(df.head())

print("Shape of the dataset:", df.shape)
```
