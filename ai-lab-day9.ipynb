```python
#Develop a python program to implement a Rule-Based Expert System for Student Performance Evaluation. The system should take inputs such as attendance, marks, assignment performance, and internal assessment and use IF THEN rules to classify the student as Excellent, Good, Average, or Poor. 

# Rule-Based Expert System for Student Performance Evaluation

print("   STUDENT PERFORMANCE EVALUATION SYSTEM")

# Taking input from the user
attendance = float(input("Enter Attendance Percentage: "))
marks = float(input("Enter Marks Percentage: "))
assignment = float(input("Enter Assignment Performance percentage: "))
internal = float(input("Enter Internal Assessment percentage: "))

# Rule-based classification

# Rule 1: Excellent
if (attendance >= 90 and marks >= 85 and
        assignment >= 85 and internal >= 85):

    performance = "Excellent"
    reason = "The student has excellent attendance, marks, assignments, and internal assessment."

# Rule 2: Good
elif (attendance >= 75 and marks >= 70 and
      assignment >= 70 and internal >= 70):

    performance = "Good"
    reason = "The student has good overall academic performance."

# Rule 3: Average
elif (attendance >= 60 and marks >= 50 and
      assignment >= 50 and internal >= 50):

    performance = "Average"
    reason = "The student has satisfactory performance but needs some improvement."

# Rule 4: Poor
else:

    performance = "Poor"
    reason = "The student needs significant improvement in academic performance."

# Displaying the result
print("EVALUATION RESULT")

print("Attendance          :", attendance, "%")
print("Marks               :", marks, "%")
print("Assignment          :", assignment, "%")
print("Internal Assessment :", internal, "%")

print("Performance Level   :", performance)
print("Reason              :", reason)
```

```python
# Rule-Based Expert System for Medical Diagnosis

print("       MEDICAL DIAGNOSIS EXPERT SYSTEM")

# Taking symptoms as input
fever = input("Do you have fever? (yes/no): ").lower()
cough = input("Do you have cough? (yes/no): ").lower()
headache = input("Do you have headache? (yes/no): ").lower()
sore_throat = input("Do you have sore throat? (yes/no): ").lower()

# Rule-based diagnosis

# Rule 1
if fever == "yes" and cough == "yes" and headache == "yes" and sore_throat == "yes":
    disease = "Flu"
    reason = "Fever, cough, headache, and sore throat are commonly associated with flu."

# Rule 2
elif fever == "yes" and cough == "yes" and sore_throat == "yes":
    disease = "Common Cold"
    reason = "Fever, cough, and sore throat may occur with a common cold."

# Rule 3
elif fever == "yes" and headache == "yes":
    disease = "Possible Viral Infection"
    reason = "Fever and headache can occur with several viral infections."

# Rule 4
elif cough == "yes" and sore_throat == "yes":
    disease = "Possible Respiratory Infection"
    reason = "Cough and sore throat can occur with respiratory infections."

# Rule 5
elif headache == "yes":
    disease = "Possible Headache/Minor Illness"
    reason = "Headache alone can have many possible causes."

# Rule 6
else:
    disease = "No Specific Disease Identified"
    reason = "The entered symptoms do not match the available rules."

# Displaying the result
print("              DIAGNOSIS RESULT")

print("Fever       :", fever)
print("Cough       :", cough)
print("Headache    :", headache)
print("Sore Throat :", sore_throat)

print("Possible Disease :", disease)
print("Reason           :", reason)

print("This result is for educational purposes only.")
```
